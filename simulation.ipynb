# Simulation on MNIST Dataset

## 1. PCA

```python
import tensorflow as tf
import numpy as np

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
X_train = np.array(X_train).flatten().reshape(60000, 28 * 28) / 255.0
y_train = np.array(y_train)
X_test = np.array(X_test).flatten().reshape(10000, 28 * 28) / 255.0
y_test = np.array(y_test)
```

```python
from pca import PCA

pca = PCA(n_components=10)

X_transformed = pca.fit_transform(X_train)
X_reconstructed = pca.reconstruct(X_transformed)
```

```python
import matplotlib.pyplot as plt

image = X_train[3].reshape(28, 28)
image_reconstructed = X_reconstructed[3].reshape(28, 28)

plt.imshow(image)
plt.show()

plt.imshow(image_reconstructed)
plt.show()
```
